```python
from pathlib import Path
import sys

project_root = Path.cwd()
for candidate in (project_root, project_root.parent, project_root.parent.parent):
    if (candidate / 'src').exists():
        project_root = candidate
        break

src_path = project_root / 'src'
if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))
```

```python
from grapheme_kit import Graphemizer

text = 'जिज्ञासा'
g = Graphemizer(text)

print('Input:', text)
print('Graphemes:', g.graphemes)
print('Count:', len(g))
print('As a list:', list(g))
```

```python
examples = [
    'माता',
    'お母さん',
    '어머니',
]

for item in examples:
    grapheme_list = list(Graphemizer(item))
    print(f'{item} -> {grapheme_list}')
```
